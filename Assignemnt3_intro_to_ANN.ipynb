```python
(x_train,y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
```

```python
print(x_train.shape, y_train.shape, x_test.shape, y_test.shape)
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
print(x_train.min(), x_train.max())
```

```python
def build_model(speciifc_nodes, num_layers =1, opti = 'adam'):
    model_net = keras.Sequential()
    model_net.add(layers.Flatten(input_shape=(28,28)))

    for i in range(num_layers):
        model_net.add(layers.Dense(speciifc_nodes, activation = 'relu'))

    model_net.add(layers.Dense(10, activation = 'softmax'))

    model_net.compile(
        optimizer = opti,
        loss = 'sparse_categorical_crossentropy',
        metrics = ['accuracy']
    )

    return model_net

build_model(128).summary()
build_model(128, num_layers = 2).summary()
```

```python
def experiment_run(run_name, different_nodes, num_of_layers, opti, batchs, epochs_size):
    with mlflow.start_run(run_name = run_name):
        mlflow.log_param('num_hidden_layers', num_of_layers)
        mlflow.log_param('nodes_per_layer', different_nodes)
        mlflow.log_param('optimizer', opti)
        mlflow.log_param('batch_size', batchs)

        model = build_model(different_nodes, num_layers = num_of_layers, opti= opti)
        history_of_model = model.fit(x_train, y_train, 
                                     epochs = epochs_size, batch_size = batchs, 
                                     validation_split = 0.1)

    print(run_name, 'final val_accuracy', history_of_model.history['val_accuracy'][-1])
    return history_of_model
```

```python
experiment_run_one = experiment_run('baseline_model', different_nodes = 128, num_of_layers = 1, 
                              opti = 'adam', batchs= 64, epochs_size = 50)
```
